### Exercise 3: Creating the Bell State ($|\Phi^+\rangle$)

**Theoretical Context:**
An entangled state is one where the state of each particle cannot be described independently. The Bell State $|\Phi^+\rangle$ is:
$$|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$$
This means if you measure qubit 0 and get `1`, you know *instantly* that qubit 1 must also be `1`.

**Your Task:**
1. Run the code below to generate the state.
2. **Observe the Q-Sphere:** Notice how the two dots represent correlated states (00 and 11).
3. **Check the Histogram:** Verify that $|01\rangle$ and $|10\rangle$ never appear.

```python
qc_bell = QuantumCircuit(2, 2)

# Step 1: Put qubit 0 into superposition
qc_bell.h(0)

# Step 2: Entangle qubit 1 with qubit 0 (Control: q0, Target: q1)
qc_bell.cx(0, 1)

# --- VISUALIZE THE ENTANGLEMENT ---
state = Statevector.from_instruction(qc_bell)
display(plot_state_qsphere(state))

# Step 3: Measure both qubits
qc_bell.measure([0, 1], [0, 1])
display(qc_bell.draw('mpl'))

counts = AerSimulator().run(qc_bell, shots=1000).result().get_counts()
print(f"Bell State Results: {counts}")
plot_histogram(counts)
```

### Exercise 4: Scaling Up to the GHZ State (3 Qubits)

**Motivation:**
Quantum computers like **Google Willow** rely on entangling dozens of qubits. A common multipartite state is the GHZ state:
$$|GHZ\rangle = \frac{|000\rangle + |111\rangle}{\sqrt{2}}$$

**Implications & Your Task:**
1. Run the code below.
2. **Reflect:** Every extra qubit makes the system more fragile. In real hardware, noise (Decoherence) would cause errors like `001` or `010` to appear in your results. Fighting this noise is the biggest challenge in the NISQ era.

```python
qc_ghz = QuantumCircuit(3, 3)

# Step 1: Superposition on q0
qc_ghz.h(0)

# Step 2: Chain the entanglement
qc_ghz.cx(0, 1)  # q0 controls q1
qc_ghz.cx(1, 2)  # q1 controls q2

qc_ghz.measure([0, 1, 2], [0, 1, 2])
display(qc_ghz.draw('mpl'))

counts = AerSimulator().run(qc_ghz, shots=1000).result().get_counts()
print(f"GHZ State Results: {counts}")
plot_histogram(counts)
```
